### 16.Programar

```python
# Sistema de Programadores - POO

class Programador:

    def __init__(self):
        self.lista = []

    def agregar_programador(self):

        datos = {
            "Nombre": input("Nombre: "),
            "Lenguaje": input("Lenguaje: "),
            "GitHub": input("GitHub: "),
            "Proyectos": int(input("Proyectos: ")),
            "Experiencia": int(input("Experiencia: ")),
            "Entorno": input("Entorno: "),
            "Nivel": input("Nivel: ")
        }

        self.lista.append(datos)

        print("Agregado")

    def listar_programadores(self):

        if len(self.lista) == 0:
            print("No hay datos")

        else:

            print("\n{:<10} {:<10} {:<12} {:<10}".format(
                "Nombre", "Lenguaje", "Experiencia", "Nivel"
            ))

            print("-" * 45)

            for p in self.lista:

                print("{:<10} {:<10} {:<12} {:<10}".format(
                    p["Nombre"],
                    p["Lenguaje"],
                    p["Experiencia"],
                    p["Nivel"]
                ))

                # IA SIMPLE
                if p["Experiencia"] >= 5:
                    print("Experto")

                elif p["Experiencia"] >= 2:
                    print("Intermedio")

                else:
                    print("Principiante")

    def actualizar_programador(self):

        nombre = input("Nombre a actualizar: ")

        for p in self.lista:

            if p["Nombre"] == nombre:
                p["Lenguaje"] = input("Nuevo lenguaje: ")
                print("Actualizado")

    def borrar_programador(self):

        nombre = input("Nombre a borrar: ")

        for p in self.lista:

            if p["Nombre"] == nombre:
                self.lista.remove(p)
                print("🗑 Eliminado")


# OBJETO
sistema = Programador()

# MENU
while True:

    print("\n1.Agregar")
    print("2.Listar")
    print("3.Actualizar")
    print("4.Borrar")
    print("5.Salir")

    op = input("Opcion: ")

    if op == "1":
        sistema.agregar_programador()

    elif op == "2":
        sistema.listar_programadores()

    elif op == "3":
        sistema.actualizar_programador()

    elif op == "4":
        sistema.borrar_programador()

    elif op == "5":
        print("Fin del programa...")
        break

    else:
        print("Opcion incorrecta")
```
